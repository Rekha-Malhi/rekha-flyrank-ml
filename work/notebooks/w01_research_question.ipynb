## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

REPO = "rekha-flyrank-ml"  # apni repo ka asli naam
URL = f"https://raw.githubusercontent.com/Rekha-Malhi/{REPO}/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(URL)

print("Total rows:", len(df))
print("Number of clients:", df["client_id"].nunique())
print("Pages with declining trend (%):", round((df["trend_direction"] == "down").mean() * 100, 1))
print("Median impressions (90 days):", df["impressions_90d"].median())
```

- The starter dataset has **30,000** pages from **32** clients, so there is plenty of content and manual review of every page is not realistic.
- About **54.2%** of pages are labeled with a declining trend. This label is a simple bucket from the current window, not a future outcome, so it is a rough starting signal only.
- The median page has **731** impressions in 90 days. This is above the 500-impression level used in the starter rules for "visible" pages, which suggests that many pages have enough volume to matter for a review queue.

These numbers suggest Lane 2 is worth my next 7 weeks: there are many pages, a large share show a declining signal, and many have enough visibility that a reviewer's time could be well spent on a ranked list.
